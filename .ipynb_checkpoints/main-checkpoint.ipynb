## Univariate analysis

```python
# Numeric features

numeric_cols = [
    "Age",
    "Avg_Daily_Usage_Hours",
    "Sleep_Hours_Per_Night",
    "Mental_Health_Score",
    "Conflicts_Over_Social_Media",
    "Addicted_Score"
]

for col in numeric_cols:
    plt.figure(figsize=(7,4))
    sns.histplot(df[col], kde=True, bins=20, color="skyblue", edgecolor="black")
    plt.title(f"Histogram & KDE of {col}", fontsize=14)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
# Categorical features 1

# checking unique values

categorical_cols = [
    "Gender", 
    "Academic_Level", 
    "Country", 
    "Most_Used_Platform", 
    "Affects_Academic_Performance", 
    "Relationship_Status"
]

# Loop through and print unique values
for col in categorical_cols:
    print(f"--- {col} ---")
    print(df[col].unique())
    print()

def plot_categorical_summary(df, col, top_n=10, min_freq=0, save=False, outdir="plots"):
    """
    Plot bar (counts) and pie (percent) for a categorical column.
    - Keeps top_n categories (after dropping NaNs). Others -> 'Others'.
    - Or, if min_freq > 0, anything below min_freq is grouped as 'Others'.
    """
    s = df[col].copy()
    s = s.fillna("Missing").astype(str)
    vc = s.value_counts(dropna=False)

    # Option A: threshold by frequency
    if min_freq > 0:
        major = vc[vc >= min_freq]
        others_count = vc[vc < min_freq].sum()
    else:
        # Option B: keep top_n
        major = vc.iloc[:top_n]
        others_count = vc.iloc[top_n:].sum()

    if others_count > 0:
        major["Others"] = others_count

    # Plot
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    colors = sns.color_palette("Set2", len(major))

    # --- Bar (counts)
    x = np.arange(len(major))
    axes[0].bar(x, major.values, color=colors)
    axes[0].set_xticks(x)
    axes[0].set_xticklabels(major.index, rotation=45, ha="right")
    axes[0].set_title(f"{col} — Counts")
    axes[0].set_ylabel("Count")

    # value labels
    for i, v in enumerate(major.values):
        axes[0].text(i, v + max(major.values)*0.02, str(int(v)), ha="center", va="bottom", fontsize=9)

    # --- Pie (percent)
    major.plot.pie(
        autopct="%1.1f%%",
        colors=colors,
        startangle=90,
        wedgeprops={"edgecolor": "black"},
        ax=axes[1]
    )
    axes[1].set_ylabel("")
    axes[1].set_title(f"{col} — Percent")

    plt.tight_layout()

    if save:
        import os
        os.makedirs(outdir, exist_ok=True)
        plt.savefig(f"{outdir}/cat_{col}.png", dpi=150, bbox_inches="tight")
    plt.show()

# ---------- Clean loop with per-column settings ----------
categorical_cols = [
    "Gender", 
    "Academic_Level", 
    "Country", 
    "Most_Used_Platform", 
    "Affects_Academic_Performance", 
    "Relationship_Status"
]

# default top_n=10, but customize specific columns
topn_map = defaultdict(lambda: 10, {
    "Most_Used_Platform": 8,
    "Country": 12  # often many categories; bump to 12
})

for col in categorical_cols:
    plot_categorical_summary(df, col, top_n=topn_map[col], min_freq=0, save=False)
```

## Bivariate analysis

### Numerical vs Numerical
- Age vs Avg_Daily_Usage_Hours
- Avg_Daily_Usage_Hours vs Sleep_Hours_Per_Night
- Mental_Health_Score vs Conflicts_Over_Social_Media
- Addicted_Score vs Mental_Health_Score

```python
def numerical_vs_numerical_analysis(df, var1, var2, regression=False, save_plot=False, plot_path=None):
    """
    Perform Bivariate Analysis for two numerical variables.
    
    Parameters:
    - df: DataFrame containing the data
    - var1: First numerical variable (string)
    - var2: Second numerical variable (string)
    - regression: Boolean flag to add regression line to scatter plot (default: False)
    - save_plot: Boolean flag to save the plot as an image file (default: False)
    - plot_path: Path to save the plot (if save_plot is True)
    
    Returns:
    - Pearson correlation, Spearman correlation, Scatter plot with optional regression line, and Heatmap
    """
    
    # Check if the variables exist in the dataframe
    if var1 not in df.columns or var2 not in df.columns:
        raise ValueError(f"Both {var1} and {var2} must be columns in the dataframe.")

    # Extract the two variables for analysis
    x = df[var1]
    y = df[var2]
    
    # 1. Scatter Plot
    plt.figure(figsize=(8, 6))
    plt.scatter(x, y, alpha=0.6, color='b', label=f'{var1} vs {var2}')
    plt.title(f'Scatter Plot: {var1} vs {var2}')
    plt.xlabel(var1)
    plt.ylabel(var2)
    
    # Add regression line if required
    if regression:
        # Fit a linear regression model
        reg = LinearRegression()
        reg.fit(x.values.reshape(-1, 1), y)
        y_pred = reg.predict(x.values.reshape(-1, 1))
        
        # Plot the regression line
        plt.plot(x, y_pred, color='red', label='Regression Line')
    
    plt.legend()
    plt.grid(True)
    
    # Save plot if required
    if save_plot:
        if plot_path:
            plt.savefig(f"{plot_path}/scatter_plot_{var1}_{var2}.png")
        else:
            plt.savefig(f"scatter_plot_{var1}_{var2}.png")
    
    plt.show()
    
    # 2. Pearson Correlation
    pearson_corr, _ = pearsonr(x, y)
    print(f"\nPearson Correlation (r) between {var1} and {var2}: {pearson_corr:.4f}")
    
    # 3. Spearman Correlation
    spearman_corr, _ = spearmanr(x, y)
    print(f"Spearman Correlation (ρ) between {var1} and {var2}: {spearman_corr:.4f}")
    
    # 4. Correlation Matrix (Heatmap)
    corr_matrix = df[[var1, var2]].corr()
    print(f"\nCorrelation Matrix:\n{corr_matrix}\n")
    
    # Heatmap of the correlation matrix
    plt.figure(figsize=(6, 5))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt='.2f')
    plt.title(f'Correlation Heatmap: {var1} vs {var2}')
    
    # Save heatmap if required
    if save_plot:
        if plot_path:
            plt.savefig(f"{plot_path}/heatmap_{var1}_{var2}.png")
        else:
            plt.savefig(f"heatmap_{var1}_{var2}.png")
    
    plt.show()
```

```python
# Age vs Avg_Daily_Usage_Hours

numerical_vs_numerical_analysis(df, 'Age', 'Avg_Daily_Usage_Hours', regression=True, save_plot=True, plot_path='./plots/bivariate_analysis/numerical_vs_numerical')

# the Age and Avg_Daily_Usage_Hours appear to have a
# very weak negative relationship, which means there is little to no meaningful correlation between them. 
# The small negative correlation values (-0.11 and -0.10) suggest that the relationship is weak and likely
# insignificant for predicting behavior in this dataset.
```

```python
# Avg_Daily_Usage_Hours vs Sleep_Hours_Per_Night

numerical_vs_numerical_analysis(df, 'Avg_Daily_Usage_Hours', 'Sleep_Hours_Per_Night', regression=True, save_plot=True, plot_path='./plots/bivariate_analysis/numerical_vs_numerical')

# The data shows a strong negative relationship between average daily usage hours and sleep hours per
# night. Students who tend to use platforms more each day generally get fewer hours of sleep at night. This
# suggests that excessive use of platforms may negatively impact sleep patterns.
```

```python
# Mental_Health_Score vs Conflicts_Over_Social_Media

numerical_vs_numerical_analysis(df, 'Mental_Health_Score', 'Conflicts_Over_Social_Media', regression=True, save_plot=True, plot_path='./plots/bivariate_analysis/numerical_vs_numerical')

# The data reveals a strong inverse relationship between mental health and conflicts over social media.
# Individuals with lower mental health scores tend to experience more conflicts related to social media, and
# vice versa. However, this correlation does not necessarily imply causality. Other variables could contribute
# to the observed relationship, and further research would be necessary to confirm any causal links.
```

```python
# Addicted_Score vs Mental_Health_Score

numerical_vs_numerical_analysis(df, 'Addicted_Score', 'Mental_Health_Score', regression=True, save_plot=True, plot_path='./plots/bivariate_analysis/numerical_vs_numerical')

# The data shows a very strong negative relationship between social media addiction and mental health.
# Individuals with higher addiction to platforms tend to report poorer mental health. However, as with
# previous analyses, correlation does not imply causality. While the relationship is clear, further research is
# needed to explore if addiction causes poorer mental health or if other factors are contributing to both.
```

### Numerical vs Categorical
- Age vs Academic_Level
- Avg_Daily_Usage_Hours vs Affects_Academic_Performance
- Sleep_Hours_Per_Night vs Relationship_Status

```python
def numerical_vs_categorical_analysis(df, num_var, cat_var, save_plot=False, plot_path='./plots/bivariate_analysis/numerical_vs_categorical'):
    """
    Perform Bivariate Analysis for Numerical vs Categorical variables.
    
    Parameters:
    - df: DataFrame containing the data
    - num_var: Numerical variable (string)
    - cat_var: Categorical variable (string)
    - save_plot: Boolean flag to save the plot as an image file (default: False)
    - plot_path: Path to save the plot (if save_plot is True)
    
    Returns:
    - ANOVA or T-test results, Box plot with the distribution of the numerical variable across categories
    """
    
    # Check if the variables exist in the dataframe
    if num_var not in df.columns or cat_var not in df.columns:
        raise ValueError(f"Both {num_var} and {cat_var} must be columns in the dataframe.")
    
    # 1. Box Plot
    plt.figure(figsize=(8, 6))
    sns.boxplot(x=cat_var, y=num_var, data=df)
    plt.title(f'Box Plot: {num_var} vs {cat_var}')
    plt.xlabel(cat_var)
    plt.ylabel(num_var)
    
    # Save plot if required
    if save_plot:
        if not os.path.exists(plot_path):
            os.makedirs(plot_path)
        plt.savefig(f"{plot_path}/boxplot_{num_var}_{cat_var}.png")
    
    plt.show()
    
    # 2. ANOVA (Analysis of Variance) test
    # Check if the categorical variable has more than two groups
    if df[cat_var].nunique() > 2:
        # Perform ANOVA
        group_data = [df[df[cat_var] == category][num_var].values for category in df[cat_var].unique()]
        f_stat, p_val = stats.f_oneway(*group_data)
        print(f"\nANOVA: F-statistic = {f_stat:.4f}, p-value = {p_val:.4f}")
        if p_val < 0.05:
            print(f"Result: There is a significant difference between the groups of {cat_var}.")
        else:
            print(f"Result: There is no significant difference between the groups of {cat_var}.")
    
    # 3. T-test (if only two categories in the categorical variable)
    else:
        # Perform T-test for two independent samples
        group1 = df[df[cat_var] == df[cat_var].unique()[0]][num_var]
        group2 = df[df[cat_var] == df[cat_var].unique()[1]][num_var]
        t_stat, p_val = stats.ttest_ind(group1, group2)
        print(f"\nT-test: t-statistic = {t_stat:.4f}, p-value = {p_val:.4f}")
        if p_val < 0.05:
            print(f"Result: There is a significant difference between the groups of {cat_var}.")
        else:
            print(f"Result: There is no significant difference between the groups of {cat_var}.")
    
    # 4. Display the count of each category for additional insights
    category_counts = df[cat_var].value_counts()
    print(f"\nCategory counts for {cat_var}:\n{category_counts}\n")
```

```python
# Age vs Academic_Level

numerical_vs_categorical_analysis(df, 'Age', 'Academic_Level', save_plot=True, plot_path='./plots/bivariate_analysis/numerical_vs_categorical')
```

```python
# Avg_Daily_Usage_Hours vs Affects_Academic_Performance

numerical_vs_categorical_analysis(df, 'Avg_Daily_Usage_Hours', 'Affects_Academic_Performance', save_plot=True, plot_path='./plots/bivariate_analysis/numerical_vs_categorical')

# The T-test confirms a significant difference in social media usage between students who feel it affects their academic performance and those who don’t.
# The imbalance in categories (453 in "Yes" vs 252 in "No") is a valid concern and should be considered when interpreting the results.
# Further research, perhaps with more balanced categories or controlling for other factors (e.g., stress, study habits), would be necessary to draw stronger conclusions.
```

```python
# Sleep_Hours_Per_Night vs Relationship_Status

numerical_vs_categorical_analysis(df, 'Sleep_Hours_Per_Night', 'Relationship_Status', save_plot=True, plot_path='./plots/bivariate_analysis/numerical_vs_categorical')

# The ANOVA test indicates a statistically significant difference in sleep hours between the three
# relationship statuses. However, it's important to consider that the "Complicated" category is much smaller
# than the other two, which could affect the robustness of the results. The small sample size could lead to
# greater variability and less statistical power in detecting true differences within this group.
```

### Categorical vs Categorical
- Gender vs Academic_Level
- Most_Used_Platform vs Affects_Academic_Performance
- Country vs Affects_Academic_Performance
- Mental_Health_Score vs Relationship_Status

```python
def categorical_vs_categorical_analysis(df, cat_var1, cat_var2, top_n=None, save_plot=False, plot_path='./plots/bivariate_analysis/categorical_vs_categorical'):
    """
    Perform Bivariate Analysis for Categorical vs Categorical variables.
    
    Parameters:
    - df: DataFrame containing the data
    - cat_var1: First categorical variable (string)
    - cat_var2: Second categorical variable (string)
    - top_n: Number of top categories to plot (default: None to show all)
    - save_plot: Boolean flag to save the plot as an image file (default: False)
    - plot_path: Path to save the plot (if save_plot is True)
    
    Returns:
    - Chi-square test result, cross-tabulation, stacked bar chart
    """
    
    # Check if the variables exist in the dataframe
    if cat_var1 not in df.columns or cat_var2 not in df.columns:
        raise ValueError(f"Both {cat_var1} and {cat_var2} must be columns in the dataframe.")
    
    # 1. Cross-tabulation (Contingency Table)
    contingency_table = pd.crosstab(df[cat_var1], df[cat_var2], margins=True)
    print(f"\nContingency Table (Cross-tabulation):\n{contingency_table}\n")
    
    # 2. Chi-Square Test of Independence
    chi2_stat, p_val, dof, expected = chi2_contingency(contingency_table.iloc[:-1, :-1])  # exclude 'All' row and column
    print(f"\nChi-Square Test of Independence:\nChi2 Stat = {chi2_stat:.4f}, p-value = {p_val:.4f}")
    
    if p_val < 0.05:
        print(f"Result: There is a significant relationship between {cat_var1} and {cat_var2}.\n")
    else:
        print(f"Result: There is no significant relationship between {cat_var1} and {cat_var2}.\n")
    
    # 3. Stacked Bar Chart (visualizing the relationship)
    plt.figure(figsize=(12, 6))
    
    # If top_n is provided, only select the top N categories
    if top_n:
        top_categories = df[cat_var1].value_counts().nlargest(top_n).index
        df_filtered = df[df[cat_var1].isin(top_categories)]
    else:
        df_filtered = df
    
    # Create the cross-tabulation for filtered data
    df_grouped = df_filtered.groupby([cat_var1, cat_var2]).size().unstack(fill_value=0)
    
    # Plot the stacked bar chart
    df_grouped.plot(kind='bar', stacked=True, figsize=(12, 6), color=sns.color_palette("Set3", len(df_grouped.columns)))
    plt.title(f'Stacked Bar Chart: {cat_var1} vs {cat_var2}')
    plt.xlabel(cat_var1)
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    
    # Save plot if required
    if save_plot:
        if not os.path.exists(plot_path):
            os.makedirs(plot_path)
        plt.savefig(f"{plot_path}/stacked_bar_{cat_var1}_{cat_var2}.png")
    
    plt.show()
```

```python
# Gender vs Academic_Level

categorical_vs_categorical_analysis(df, 'Gender', 'Academic_Level', save_plot=True, plot_path='./plots/bivariate_analysis/categorical_vs_categorical')
```

```python
# Most_Used_Platform vs Affects_Academic_Performance

categorical_vs_categorical_analysis(df, 'Most_Used_Platform', 'Affects_Academic_Performance', save_plot=True, plot_path='./plots/bivariate_analysis/categorical_vs_categorical')

# The statistical tests confirm that the type of platform used significantly relates to whether a student feels
# social media affects their academic performance. Platforms like Instagram and TikTok are likely to be
# more distracting and thus more likely to negatively impact academic performance, while platforms like
# LinkedIn and YouTube are more likely to be used for educational purposes, leading to less negative impact

# Chi-Square test only shows that there is a significant relationship between these two variables, but it does not prove causality.
# Further research, possibly including longitudinal studies, would be needed to explore the causal relationship between social media usage and academic performance.
```

```python
# Country vs Affects_Academic_Performance

categorical_vs_categorical_analysis(df, 'Country', 'Affects_Academic_Performance', top_n=20, save_plot=True, plot_path='./plots/bivariate_analysis/categorical_vs_categorical')

# The test confirms a significant relationship between Country and Affects Academic Performance
# meaning that students from different countries perceive the effect of social media on academic
# performance differently. However, due to imbalanced data (e.g., India having much more data than other
# countries), the interpretation should be cautious.
```

```python
# Mental_Health_Score vs Relationship_Status

categorical_vs_categorical_analysis(df, 'Mental_Health_Score', 'Relationship_Status', top_n=20, save_plot=True, plot_path='./plots/bivariate_analysis/categorical_vs_categorical')

# The analysis suggests that relationship status might influence mental health, with people in relationships
# tending to report higher mental health scores. However, causality cannot be established based on this
# data alone. There might be other factors (such as stress, social support, or time management) that
# contribute to the observed relationship.
```
